```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
words = ["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized","congartulations"]
```

```python
'''
   POS 
Noun -> n
Verb -> v
Adjective -> a
Adverb -> r
'''

print(lemmatizer.lemmatize('going',pos='n'))
print(lemmatizer.lemmatize('going',pos='v'))
print(lemmatizer.lemmatize('going',pos='a'))
print(lemmatizer.lemmatize('going',pos='r'))
```

```python
for word in words:
    print(word+"--->"+lemmatizer.lemmatize(word,pos='n'))
```

```python
for word in words:
    print(word+"--->"+lemmatizer.lemmatize(word,pos='v'))
```

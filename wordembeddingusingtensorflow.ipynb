```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'you are too good'
]
```

```python
sent
```

```python
## Define vocabulary size
voc_size = 10000
```

```python
##one hot representation for every word
one_hot_rep = [one_hot(words,voc_size) for words in sent]
one_hot_rep
```

### 'the' feature is represented using one-hot encoding, with the 1521st position set to 1 and all remaining positions set to 0.

```python
## word embedding represntation
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
'''
We need to make every sentences of equal size so we use pad_sequence and set a max sentence length
addind zeros, post or pre.
'''

sent_length = 8
embedded_docs = pad_sequences(one_hot_rep,padding='pre',maxlen = sent_length)
print(embedded_docs)
```

```python
## Feature rep
dim=10
```

```python
model=Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_length))
model.build(input_shape=(None, sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```

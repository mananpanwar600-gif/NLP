```python
words = ["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized","congartulations"]
```

## Porter Stemmer


```python
from nltk.stem import PorterStemmer
stemming = PorterStemmer()
```

```python
for word in words:
    print(word+"---->"+stemming.stem(word))
```

## Regexp Stemmer

```python
from nltk.stem import RegexpStemmer
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

```python
reg_stemmer1 = RegexpStemmer('ing|s$|e$|able$', min=4)
```

```python
reg_stemmer1.stem('ingeating')
```

## Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
snowball = SnowballStemmer("english")
```

```python
for word in words:
    print(word+"--->"+snowball.stem(word))
```

## Porter vs Snowball

```python
stemming.stem('fairly'), stemming.stem('sportingly')
```

```python
snowball.stem('fairly'),snowball.stem('sportingly')
```

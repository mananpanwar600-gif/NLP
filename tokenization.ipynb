```python
corpus = """Hello Welcome, I am MSP. This is MSP's code.
Please refer to this code! to gain knowledge on Tokenization."""
```

```python
print(corpus)
```

# Tokenization

### Pragraph--->Sentences

```python
from nltk.tokenize import sent_tokenize
documents=sent_tokenize(corpus)
```

```python
documents
```

```python
type(documents)
```

### Paragraph---> Words

```python
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

### Sentence--->Words

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
### How can we take this apostropy ('s) as two different tokens.
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

### Tree Bank Word Tokenizer

Full Stop is not treated as a seprate token and only the last full stop where the paragraph ends is treated as a seprate token

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)

```
